```python
x = 10
print(x)
print(type(x))

#float(var) -> convert to float
x = float(x)
print(x)
print(type(x))
```

```python
y = 5.7
print(y)
print(type(y))

#int(var) -> convert to integer
y = int(y)
print(y)
print(type(y))
```

```python
z = 15
print(z)
print(type(z))

#str(var) -> convert to string
z = str(z)
print(z)
print(type(z))
```

```python
a = 15 #int
b = "15" #str

print(a)
print(b)
print(type(a))
print(type(b))
```

```python
age = input("Enter your age: ")
print(age)
print(type(age)) #when using input the data type is always converted to string
```

```python
#to convert the output of input to int we can use int() function
age = int(input("Enter your age: "))
print(age)
print(type(age))
```

```python
#to convert the output of input to float we can use float() function
height = float(input("Enter your age: "))
print(height)
print(type(height))
```
